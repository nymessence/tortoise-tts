```python
%%writefile generate_audio.py

import os
import sys
import logging
import torch
import torchaudio
import torch_xla.distributed.xla_multiprocessing as xmp
import time
import threading
from cloud_tpu_client import Client
from pathlib import Path

# Ensure Tortoise fork is in path
sys.path.insert(0, '/tmp/tortoise-tts')

from tortoise.api import run_generation_local, run_generation_for_spawn
from tortoise.utils.generate_tts_core import preprocess_script, stitch_audio
from tortoise.binaural import add_hemi_sync, BINAURAL_BEATS

# Configure logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - [%(process)d] %(message)s')

# 🧠 Global flag to control TPU monitoring thread
_tpu_monitor_stop = threading.Event()

# ➕ GET GIT SHA — Know exactly what code is running
def get_git_sha(repo_path="/tmp/tortoise-tts"):
    try:
        import subprocess
        result = subprocess.run(
            ['git', 'rev-parse', '--short', 'HEAD'],
            cwd=repo_path,
            capture_output=True,
            text=True,
            check=True
        )
        return result.stdout.strip()
    except Exception:
        return "unknown"

GIT_SHA = get_git_sha()
logging.info(f"🦎 TORTOISE-TTS FORK git-{GIT_SHA} LOADED — Micro-batching + List Safety")

# ✅ Validate functions in correct module
required_functions = {
    'run_generation_for_spawn': 'tortoise.utils.generate_tts_core',
    'run_generation_local': 'tortoise.utils.generate_tts_core',
    'preprocess_script': 'tortoise.utils.generate_tts_core',
    'stitch_audio': 'tortoise.utils.generate_tts_core',
}

for func_name, module_name in required_functions.items():
    try:
        module = sys.modules[module_name]
        if not hasattr(module, func_name):
            raise AttributeError(f"Function '{func_name}' not found in module '{module_name}'")
    except (KeyError, AttributeError) as e:
        logging.critical(f"❌ {e}. Did you deploy the right version?")
        sys.exit(1)

logging.info("✅ All required functions are available.")

# 📊 TPU Memory Monitor
def monitor_tpu_memory(interval=30):
    """
    Monitors TPU memory usage in a background thread.
    Logs usage every `interval` seconds until stop event is set.
    """
    try:
        c = Client()
        logging.info("📊 TPU Memory Monitor started...")
        while not _tpu_monitor_stop.is_set():
            try:
                usage = c.get_memory_usage()
                total = sum(usage)
                avg = total / len(usage) if usage else 0
                usage_str = ", ".join([f"{u:.0f}MB" for u in usage])
                logging.info(f"GMEM ➤ Total: {total:.0f}MB | Avg/Core: {avg:.0f}MB | Per Core: [{usage_str}]")
            except Exception as e:
                logging.warning(f"⚠️ Failed to get TPU memory usage: {e}")
            time.sleep(interval)
    except Exception as e:
        logging.error(f"GMEM ➤ Monitor thread crashed: {e}")

# 🛡️ Safe Wrapper for TPU Generation
def run_generation_for_spawn_safe(index, process_args_list):
    """
    Wraps the original run_generation_for_spawn with error handling.
    If a line fails, saves silent audio and continues.
    """
    flags = process_args_list[index]
    output_dir = Path(flags['output_dir'])
    start_idx = flags['start_idx']
    end_idx = flags['end_idx']
    rank = flags['rank']

    # ➤ Set up per-core logging to file
    log_path = output_dir / f"core_{rank}_log.txt"
    file_handler = logging.FileHandler(log_path, mode='a')
    file_handler.setFormatter(logging.Formatter('%(asctime)s - %(levelname)s - %(message)s'))
    logger = logging.getLogger()
    logger.addHandler(file_handler)
    logging.info(f"[Core {rank}] Log file initialized: {log_path}")

    # Load lines
    try:
        with open(flags['lines_file'], 'r', encoding='utf-8') as f:
            all_lines = [line.strip() for line in f if line.strip()]
    except Exception as e:
        logging.error(f"[Core {rank}] Failed to load lines: {e}")
        all_lines = [""] * (end_idx - start_idx)

    # Silent fallback
    sample_rate = flags.get('sample_rate', 22050)
    silent_audio = torch.zeros(1, sample_rate)

    # Try normal generation
    try:
        logging.info(f"[Core {rank}] Starting generation for lines [{start_idx} - {end_idx})...")
        run_generation_for_spawn(index, process_args_list)
        logging.info(f"[Core {rank}] Generation completed successfully.")
    except Exception as main_e:
        if "'list' object has no attribute" in str(main_e):
            logging.critical(f"[Core {rank}] 🚨 YOU ARE RUNNING AN OUTDATED VERSION — Please reinstall tortoise-tts fork with list-safety fix!")
        logging.error(f"[Core {rank}] 💥 Main generation failed: {main_e}")
        logging.info(f"[Core {rank}] 🛡️ Switching to safe fallback mode...")

    # ✅ Safety check + heartbeat
    start_time = time.time()
    last_heartbeat = start_time

    for i in range(start_idx, end_idx):
        current_time = time.time()
        if current_time - last_heartbeat > 60:
            elapsed = current_time - start_time
            remaining = end_idx - i
            logging.info(f"[Core {rank}] ❤️ Heartbeat: {elapsed:.0f}s elapsed, {remaining} lines remaining...")
            last_heartbeat = current_time

        output_path = output_dir / f"line_{i+1:04d}.wav"
        if not output_path.exists():
            line = all_lines[i] if i < len(all_lines) else "UNKNOWN"
            # ➕ "FIX ME" for easy grep later
            logging.warning(f"[Core {rank}] 🚫 Missing or failed line {i+1}. Saved 1s silence. FIX ME: {line[:50]}...")
            try:
                save_audio_fallback(output_path, silent_audio, sample_rate)
                logging.info(f"[Core {rank}] 🎧 Saved silent fallback: {output_path}")
            except Exception as save_e:
                logging.error(f"[Core {rank}] ❌ Failed to save fallback: {save_e}")

    # Clean up
    logger.removeHandler(file_handler)
    file_handler.close()
    logging.info(f"[Core {rank}] ✅ Chunk [{start_idx}-{end_idx}) complete with error recovery.")


# ✅ FUTURE-PROOF AUDIO SAVING — Uses torchcodec.encoders.AudioEncoder
def save_audio_fallback(path, tensor, sample_rate):
    """
    Saves audio tensor to file using torchcodec.encoders.AudioEncoder (recommended by PyTorch).
    Falls back to torchaudio.save if torchcodec is unavailable.
    """
    try:
        from torchcodec.encoders import AudioEncoder

        # Normalize to [-1, 1] if needed
        if tensor.abs().max() > 1.0:
            tensor = tensor / tensor.abs().max()

        # Ensure shape [channels, samples]
        if tensor.dim() == 1:
            tensor = tensor.unsqueeze(0)

        # Create encoder — matches PyTorch docs
        encoder = AudioEncoder(tensor, sample_rate=sample_rate)

        # Save to file — format inferred from extension
        # Using 192 kbps for high quality
        encoder.to_file(str(path), bit_rate=192000)

        logging.debug(f"✅ Saved with torchcodec: {path}")
        return

    except ImportError:
        logging.debug("torchcodec not available, falling back to torchaudio")
    except Exception as e:
        logging.warning(f"torchcodec failed: {e}, falling back to torchaudio")

    # Fallback to torchaudio
    torchaudio.save(str(path), tensor, sample_rate=sample_rate)
    logging.debug(f"✅ Saved with torchaudio: {path}")


# 🎯 MAIN ENTRY POINT
def run_tts():
    OUTPUT_BASE_DIR = Path("/kaggle/working")
    OUT_DIR = OUTPUT_BASE_DIR / "speech"
    SCRIPT_PATH = OUTPUT_BASE_DIR / "story_script.txt"
    PREPROCESSED_SCRIPT_PATH = OUT_DIR / "story_script_preprocessed.txt"
    TTS_DIR = "/tmp/tortoise-tts"

    LINES_DIR = OUT_DIR / "lines"
    STORY_AUDIO_PATH = OUT_DIR / "story.wav"
    FINAL_AUDIO_PATH = OUT_DIR / "story_with_beats.wav"

    LINES_DIR.mkdir(parents=True, exist_ok=True)

    logging.info("📝 Preprocessing the script...")
    preprocess_script(str(SCRIPT_PATH), str(PREPROCESSED_SCRIPT_PATH))

    try:
        with open(PREPROCESSED_SCRIPT_PATH, 'r', encoding='utf-8') as f:
            lines = [line.strip() for line in f if line.strip()]
            total_lines = len(lines)
    except FileNotFoundError:
        logging.error(f"Preprocessed script not found: {PREPROCESSED_SCRIPT_PATH}")
        sys.exit(1)

    # ✅ FORCE REGENERATE — Set to False to enable resume
    FORCE_REGENERATE = True  # ← 🔥 SET TO TRUE TO ALWAYS REGENERATE

    if FORCE_REGENERATE:
        logging.info("♻️ FORCE_REGENERATE enabled — ignoring existing files.")
        start_idx = 0
    else:
        # --- Resume Logic ---
        existing_files = sorted([f for f in os.listdir(LINES_DIR) if f.endswith('.wav')])
        start_idx = len(existing_files)
        logging.info(f"✅ Found {start_idx} previously generated files. Resuming from line {start_idx + 1}.")
        # --------------------

    if start_idx >= total_lines:
        logging.info("🎉 All lines already generated. Skipping generation.")
    else:
        # --- Split remaining lines across 8 TPU processes ---
        remaining_lines = total_lines - start_idx
        nprocs = 8
        chunk_size = (remaining_lines + nprocs - 1) // nprocs

        process_args = []
        for rank in range(nprocs):
            chunk_start = start_idx + rank * chunk_size
            chunk_end = min(start_idx + (rank + 1) * chunk_size, total_lines)
            if chunk_start >= chunk_end:
                logging.debug(f"Core {rank} has no lines to process.")
                continue
            process_args.append({
                'lines_file': str(PREPROCESSED_SCRIPT_PATH),
                'output_dir': str(LINES_DIR),
                'hardware': 'tpu',
                'voice': "nya",
                'preset': "ultra_fast",
                'diffusion_iterations': 16,
                'num_autoregressive_samples': 1,
                'models_dir': TTS_DIR,
                'start_idx': chunk_start,
                'end_idx': chunk_end,
                'total_lines': total_lines,
                'rank': rank,
                'batch_size': 4,
                'sample_rate': 22050,
            })
            logging.debug(f"Core {rank} assigned lines [{chunk_start} - {chunk_end})")

        # ➕ Log config summary
        config_summary = {
            "voice": process_args[0]['voice'],
            "preset": process_args[0]['preset'],
            "diffusion_iterations": process_args[0]['diffusion_iterations'],
            "batch_size": process_args[0]['batch_size'],
            "sample_rate": process_args[0]['sample_rate'],
            "total_lines": total_lines,
            "nprocs": len(process_args),
            "chunk_size": chunk_size,
        }
        logging.info(f"⚙️  CONFIG: {config_summary}")

        logging.info(f"🗣️ Starting TPU generation with {len(process_args)} processes, each handling up to {chunk_size} lines.")

        # ✅ START TPU MEMORY MONITORING (if on TPU)
        tpu_monitor_thread = None
        if 'TPU_NAME' in os.environ:
            _tpu_monitor_stop.clear()
            tpu_monitor_thread = threading.Thread(target=monitor_tpu_memory, args=(30,), daemon=True)
            tpu_monitor_thread.start()
            logging.info("GMEM ➤ TPU memory monitoring started in background thread.")

        # ➕ Start timer
        generation_start_time = time.time()

        try:
            if 'TPU_PROCESSES_COUNT' in os.environ and 'TPU_NAME' in os.environ:
                logging.info("✅ TPU detected. Starting chunked generation across 8 cores.")
                xmp.spawn(run_generation_for_spawn_safe, args=(process_args,), nprocs=len(process_args))
            else:
                logging.info("⚠️ TPU not detected. Falling back to single-process generation.")
                device_str = 'cuda' if torch.cuda.is_available() else 'cpu'
                flags = process_args[0]
                flags['hardware'] = device_str
                try:
                    run_generation_local(flags)
                except Exception as e:
                    logging.error(f"❌ Local generation failed: {e}")
                    with open(flags['lines_file'], 'r', encoding='utf-8') as f:
                        all_lines = [line.strip() for line in f if line.strip()]
                    silent_audio = torch.zeros(1, flags.get('sample_rate', 22050))
                    for i in range(flags['start_idx'], flags['end_idx']):
                        output_path = LINES_DIR / f"line_{i+1:04d}.wav"
                        if not output_path.exists():
                            save_audio_fallback(output_path, silent_audio, flags.get('sample_rate', 22050))
        except Exception as e:
            logging.error(f"❌ Audio generation orchestration failed: {e}")
            sys.exit(1)
        finally:
            # ✅ STOP TPU MONITORING
            if tpu_monitor_thread:
                _tpu_monitor_stop.set()
                tpu_monitor_thread.join(timeout=5)
                logging.info("GMEM ➤ TPU memory monitoring stopped.")

        # ➕ Log generation time
        generation_elapsed = time.time() - generation_start_time
        logging.info(f"⏱️  Total TTS generation time: {generation_elapsed:.1f} seconds")

    logging.info("✅ Audio generation complete.")

    # --- Verify all expected files exist ---
    expected_files = [f"line_{i+1:04d}.wav" for i in range(total_lines)]
    missing_files = [f for f in expected_files if not (LINES_DIR / f).exists()]

    if missing_files:
        logging.error(f"❌ Missing {len(missing_files)} files. First 5: {missing_files[:5]}")
        sys.exit(1)

    logging.info("✂️ Stitching audio segments...")
    stitch_audio(str(LINES_DIR), total_lines, str(STORY_AUDIO_PATH))

    # ✅ Verify stitched audio exists
    if not STORY_AUDIO_PATH.exists():
        logging.critical(f"❌ Stitched audio not found: {STORY_AUDIO_PATH}")
        sys.exit(1)

    logging.info("🎶 Layering binaural beats...")
    add_hemi_sync(str(STORY_AUDIO_PATH), str(FINAL_AUDIO_PATH), beat_freqs=[BINAURAL_BEATS['theta'], BINAURAL_BEATS['alpha']])

    # ✅ Verify final audio exists
    if not FINAL_AUDIO_PATH.exists():
        logging.critical(f"❌ Final audio with beats not found: {FINAL_AUDIO_PATH}")
        sys.exit(1)

    logging.info("🎉🎉🎉 AUDIO PIPELINE SUCCESS — Ready for ffmpeg 🎬")
    sys.exit(0)


if __name__ == "__main__":
    run_tts()
```
